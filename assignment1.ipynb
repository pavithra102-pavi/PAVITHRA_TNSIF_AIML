```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter number: "))
    numbers.append(num)

reverse = []

for i in range(n - 1, -1, -1):
    reverse.append(numbers[i])

print("Reversed Array:", reverse)
```

```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter number: "))
    numbers.append(num)

even = 0
odd = 0
zero = 0

for num in numbers:
    if num == 0:
        zero += 1
    elif num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
print("Zero:", zero)
```

```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter number: "))
    numbers.append(num)

positive = 0
negative = 0

for num in numbers:
    if num > 0:
        positive += num
    elif num < 0:
        negative += num

print("Positive Sum:", positive)
print("Negative Sum:", negative)
```

```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter number: "))
    numbers.append(num)

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("Array after removing duplicates:", unique)
```

```python
n = int(input("Enter N: "))

numbers = []

print("Enter", n - 1, "numbers:")

for i in range(n - 1):
    num = int(input("Enter number: "))
    numbers.append(num)

total = 0

for i in range(1, n + 1):
    total += i

for num in numbers:
    total -= num

print("Missing Number:", total)
```

```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter number: "))
    numbers.append(num)

k = int(input("Enter K: "))

for i in range(k):
    last = numbers.pop()
    numbers.insert(0, last)

print("Rotated Array:", numbers)
```

```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter number: "))
    numbers.append(num)

most = numbers[0]
max_count = 0

for num in numbers:
    count = 0

    for x in numbers:
        if num == x:
            count += 1

    if count > max_count:
        max_count = count
        most = num

print("Most Frequent Element:", most)
print("Frequency:", max_count)
```

```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter number: "))
    numbers.append(num)

max_sum = numbers[0]
current_sum = 0

for num in numbers:
    current_sum += num

    if current_sum < 0:
        current_sum = 0

    if current_sum > max_sum:
        max_sum = current_sum

print("Maximum Subarray Sum:", max_sum)
```
