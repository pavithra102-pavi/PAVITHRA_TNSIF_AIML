```python
nums = [2, 7, 11, 15]
target = 9

seen = {}

for i, num in enumerate(nums):
    complement = target - num

    if complement in seen:
        print("Indices:", [seen[complement], i])
        break

    seen[num] = i
```

```python
s = "abcabcbb"

seen = set()
left = 0
maximum = 0

for right in range(len(s)):
    while s[right] in seen:
        seen.remove(s[left])
        left += 1

    seen.add(s[right])
    maximum = max(maximum, right - left + 1)

print("Length:", maximum)
```

```python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

current = nums[0]
maximum = nums[0]

for i in range(1, len(nums)):
    current = max(nums[i], current + nums[i])
    maximum = max(maximum, current)

print("Maximum Subarray Sum:", maximum)
```

```python
intervals = [[1, 3], [2, 6], [8, 10], [9, 12]]

intervals.sort()
result = []

for start, end in intervals:
    if not result or result[-1][1] < start:
        result.append([start, end])
    else:
        result[-1][1] = max(result[-1][1], end)

print("Merged Intervals:", result)
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


head = Node(1)
second = Node(2)
third = Node(3)

head.next = second
second.next = third
third.next = second   # Creates cycle


slow = head
fast = head
cycle = False

while fast and fast.next:
    slow = slow.next
    fast = fast.next.next

    if slow == fast:
        cycle = True
        break

print("Cycle exists:", cycle)
```

```python
s = "({[]})"

stack = []
pairs = {
    ')': '(',
    '}': '{',
    ']': '['
}

valid = True

for char in s:
    if char in "({[":
        stack.append(char)
    else:
        if not stack or stack[-1] != pairs[char]:
            valid = False
            break
        stack.pop()

if stack:
    valid = False

print("Valid Parentheses:", valid)
```

```python
from collections import deque

class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


root = Node(3)
root.left = Node(9)
root.right = Node(20)
root.right.left = Node(15)
root.right.right = Node(7)


queue = deque([root])
result = []

while queue:
    level = []

    for _ in range(len(queue)):
        node = queue.popleft()
        level.append(node.value)

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)

    result.append(level)

print("Level Order:", result)
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


root = Node(3)
root.left = Node(5)
root.right = Node(1)

root.left.left = Node(6)
root.left.right = Node(2)

root.right.left = Node(0)
root.right.right = Node(8)


def lca(root, p, q):
    if root is None or root == p or root == q:
        return root

    left = lca(root.left, p, q)
    right = lca(root.right, p, q)

    if left and right:
        return root

    return left if left else right


p = root.left
q = root.right

answer = lca(root, p, q)

print("Lowest Common Ancestor:", answer.value)
```

```python
grid = [
    ['1', '1', '0', '0'],
    ['1', '0', '0', '1'],
    ['0', '0', '1', '1']
]

rows = len(grid)
cols = len(grid[0])
count = 0


def dfs(r, c):
    if r < 0 or r >= rows or c < 0 or c >= cols:
        return

    if grid[r][c] == '0':
        return

    grid[r][c] = '0'

    dfs(r + 1, c)
    dfs(r - 1, c)
    dfs(r, c + 1)
    dfs(r, c - 1)


for r in range(rows):
    for c in range(cols):
        if grid[r][c] == '1':
            count += 1
            dfs(r, c)

print("Number of Islands:", count)
```

```python
nums = [100, 4, 200, 1, 3, 2]

num_set = set(nums)
longest = 0

for num in num_set:

    if num - 1 not in num_set:

        current = num
        length = 1

        while current + 1 in num_set:
            current += 1
            length += 1

        longest = max(longest, length)

print("Longest Consecutive Sequence:", longest)
```
